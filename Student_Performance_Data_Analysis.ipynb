Student Performance Data Analysis

Project Overview:
This project explores student performance using a dataset containing students' math, reading, and writing scores, along with demographic, lunch and educational background information.
The purpose of this analysis is to identify patterns in student performance, compare scores across different subjects, and explore how factors such as gender, race/ethnicity categories, test preparation, lunch and parental education relate to academic outcomes.

Project Objectives:
The main objectives of this project are:
1. Explore the dataset: Understand its structure, variables, and overall information.
2. Analyze student scores: Calculate averages, medians, and standard deviations for math, reading, and writing.
3. Visualize score distributions: Use graphs to understand how students performed in each subject.
4. Examine relationships: Explore how scores in different subjects are related.
5. Compare academic performance: Identify students' highest and lowest scores and examine score differences.
6. Explore student characteristics: Investigate performance patterns across gender, race/ethnicity categories, test preparation status, and parental education levels.
Tools and Libraries
This project uses the following Python libraries:
- Pandas: Organizing, exploring, and analyzing the dataset.
- NumPy: Performing numerical calculations.
- Matplotlib: Creating graphs and visualizations

Dataset:
The dataset used in this project is StudentsPerformance.csv. It contains information about student scores and background characteristics.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

print("Libraries imported successfully! We can now start our data analysis project!!!")
```

```python
df = pd.read_csv("StudentsPerformance.csv")
print("Dataset loaded successfully!")
```

```python
print("Dataset preview:")
print("Dataset dimensions:", df.shape)

print("Column names:")
print(df.columns)

print("First five rows:")
print(df.head())
```

```python
print("What are the data types and structure of the dataset?")
df.info()
```

```python
print("Are there any missing values in the student performance dataset?")
missing_values = df.isnull().sum()
print("Missing values in each column: ")
print(missing_values)
```

```python
# mean using pandas
math_scores = df["math score"]
reading_scores = df["reading score"]
writing_scores = df["writing score"]

average_math = math_scores.mean()
average_reading = reading_scores.mean()
average_writing = writing_scores.mean()

print("Math:", average_math)
print("Reading:", average_reading)
print("Writing:", average_writing)

# mean using numpy

average_math_np = np.mean(math_scores)
average_reading_np = np.mean(reading_scores)
average_writing_np = np.mean(writing_scores)

print("Math:", average_math_np)
print("Reading:", average_reading_np)
print("Writing:", average_writing_np)

```

```python
print("Which subject has the highest average score?")
print("This is a bar chart of Average Student Scores by Subject")
subjects = ["Math", "Reading", "Writing"]
averages = [
        average_math,
        average_reading, 
        average_writing
]
    

colors = ["skyblue", "pink","lightgreen" ]
plt.bar(subjects, averages, color= colors, edgecolor = "black")
plt.title("Average Student Scores by Subject")
plt.xlabel("Subject")
plt.ylabel("Average Score")

plt.show()
```

```python
print("How are math scores distributed among students?")
print("This is a histogram of the distribution of Math scores")
plt.hist(math_scores, bins = 10, color="skyblue", edgecolor = "Black")
plt.title("Distribution of math scores")
plt.xlabel("Math Scores")
plt.ylabel("Number of students")
plt.show()
```

```python
print("How are reading scores distributed among students?")
print("This is a histogram of distribution of Reading scores")
plt.hist(reading_scores, bins = 10, color="pink" ,edgecolor = "Black")
plt.title("Distribution of Reading Scores")
plt.xlabel("Reading Scores")
plt.ylabel("Number of students")
plt.show()
```

```python
print("How are writing scores distributed among students?")
print("this is a histogram of the distribution of Writing scores")
plt.hist(writing_scores, bins = 10, color="lightgreen", edgecolor = "Black")
plt.title("Distribution of writing scores")
plt.xlabel("Writing Scores")
plt.ylabel("Number of students")
plt.show()
```

```python
print("Is there a relationship between students' math scores and reading scores?")
scatter_colors = ["gray"]
plt.scatter(math_scores, reading_scores, color=scatter_colors, edgecolor = "lightblue")
plt.title("Math Scores vs Reading Scores")
plt.xlabel("Math Scores")
plt.ylabel("Reading Scores")
plt.show()
```

```python
print("Is there a relationship between students' math scores and writing scores?")
scatter_colors = ["purple"]
plt.scatter(math_scores, writing_scores, color=scatter_colors, edgecolor = "lightgray")
plt.title("Math Scores vs Writing Scores")
plt.xlabel("Math Scores")
plt.ylabel("Writing Scores")
plt.show()
```

```python
print("Is there a relationship between students' writing scores and reading scores?")
scatter_colors = ["pink"]
plt.scatter(writing_scores, reading_scores, color=scatter_colors, edgecolor = "black")
plt.title("Writing Scores vs Reading Scores")
plt.xlabel("Writing Scores")
plt.ylabel("Reading Scores")
plt.show()
```

```python
print("How do the distributions of student scores compare across math, reading, and writing?")
score_data = [
    math_scores,
    reading_scores,
    writing_scores
]

plt.boxplot(score_data, tick_labels=["Math", "Reading", "Writing"])

plt.title("Score Distribution by Subject")
plt.xlabel("Subject")
plt.ylabel("Score")

plt.show()
```

```python
print("What is the correlation between students' scores in math, reading, and writing?")
correlation_matrix = df[["math score", "reading score", "writing score"]]
print(correlation_matrix.corr())
```

```python
print("How many students scored equally in reading and writing?")

equal_reading_writing = df[
    (df["reading score"] == df["writing score"])
]
print("Number of students with equal reading and writing scores:", len(equal_reading_writing))
```

```python
print("How many students scored equally in reading and math?")

equal_reading_math = df[
    (df["reading score"] == df["math score"])
]

print("Number of students with equal reading and math scores:", len(equal_reading_math))
```

```python
print("How many students scored equally in writing and math?")

equal_writing_math = df[
    (df["writing score"] == df["math score"])
]

print("Number of students with equal writing and math scores:", len(equal_writing_math))
```

```python
print("How many students scored equally in all three subjects?")

equal_scores = df[
    (df["math score"] == df["reading score"]) &
    (df["reading score"] == df["writing score"])
]

print("Number of students with equal scores:", len(equal_scores))
```

```python
print("Does the dataset contain any duplicate value?")
duplicate_count = df.duplicated().sum()
print("Number of duplicate records: ", duplicate_count)
```

```python
print("What are the minimum, maximum, and average scores in each subject?")

score_statistics = df[["math score", "reading score", "writing score"]].describe()

print(score_statistics)
```

```python
print("How can we classify students' math performance into different score categories?")
def classify_math_score(score):
  if score >= 90:
    return "Excellent"
  elif score >= 70:
    return "Good"
  elif score >= 50:
    return "Needs Improvement"
  else:
    return "Low score"
print(classify_math_score(95))
print(classify_math_score(85))
print(classify_math_score(60))
print(classify_math_score(35))

## Classify all students
df["math performance"] = df["math score"].apply(classify_math_score)

# Display the first 10 students
print("Math scores and performance categories:")
print(df[["math score", "math performance"]].head(10))
```

```python
print("What percentage of students fall into each math performance category?")

performance_percentages = df["math performance"].value_counts(normalize=True) * 100

print("\nPercentage of students in each category:")
print(performance_percentages.round(2))
```

```python
print("Preparing additional student performance columns...")

score_columns = ["math score", "reading score", "writing score"]

# Average score across all three subjects
df["average score"] = df[score_columns].mean(axis=1)

# Highest score across the three subjects
df["highest score"] = df[score_columns].max(axis=1)

# Lowest score across the three subjects
df["lowest score"] = df[score_columns].min(axis=1)

# Difference between the highest and lowest subject scores
df["score gap"] = df["highest score"] - df["lowest score"]

print("New columns created successfully!")
```

```python
print("Which students have the highest overall averages?")

top_students = df.sort_values(
    by="average score",
    ascending=False
)

print("\nTop 10 students by overall average:")
print(
    top_students[
        ["math score", "reading score", "writing score", "average score"]
    ].head(10).round(2)
)
```

```python
print("Which subject has the highest median score?")

median_scores = df[score_columns].median()
bar_colors = ["skyblue", "pink","lightgreen" ]
plt.bar(score_columns, median_scores.values, color=bar_colors, edgecolor = "black")

plt.title("Median Student Scores by Subject")
plt.xlabel("Subject")
plt.ylabel("Median Score")
plt.show()
```

```python
print("Which subject has the greatest variation in scores?")

score_std = df[score_columns].std()


colors = ["skyblue", "pink","lightgreen" ]
plt.bar(score_columns, score_std.values, color=colors, edgecolor = "black")
plt.title("Score Variation by Subject")
plt.xlabel("Subject")
plt.ylabel("Standard Deviation")
plt.show()
```

```python
print("What is the difference between each student's highest and lowest subject scores?")
score_columns = ["math score", "reading score", "writing score"]
df[["math score", "reading score", "writing score", "score gap"]].head()
```

```python
print("Graph of the distribution of highest scores among students")
plt.hist(df["highest score"], bins = 10, color = "yellow", edgecolor = "black")
plt.xlabel("Highest Score")
plt.ylabel("Frequency")
plt.title("Distribution of Highest Scores")
plt.show()
```

```python
print("Graph of the distribution of lowest scores among students")
plt.hist(df["lowest score"], bins = 10, color = "lightgreen", edgecolor = "black")
plt.xlabel("Lowest Score")
plt.ylabel("Frequency")
plt.title("Distribution of Lowest Scores")
plt.show()
```

```python
print("How many students scored above the average in each subject?")
above_average_math = df[df["math score"] > df["math score"].mean()]
above_average_reading = df[df["reading score"] > df["reading score"].mean()]
above_average_writing = df[df["writing score"] > df["writing score"].mean()]
print(f"Number of students who scored above average in math: {len(above_average_math)}")
print(f"Number of students who scored above average in reading: {len(above_average_reading)}")
print(f"Number of students who scored above average in writing: {len(above_average_writing)}")
```

```python
print("How many students scored above average in all three subjects?")
above_average_all = df[
    (df["math score"] > df["math score"].mean()) &
    (df["reading score"] > df["reading score"].mean()) &
    (df["writing score"] > df["writing score"].mean())
]
print("Number of students with above-average scores:", len(above_average_all))
```

```python
print("Graph of students who scored above average in math")
plt.hist(above_average_math["math score"], bins = 10, color = "lightcoral", edgecolor = "black")
plt.xlabel("Math Score")
plt.ylabel("Frequency")
plt.title("Distribution of Students Scoring Above Average in Math")
plt.show()
```

```python
print("Graph of students who scored above average in reading")
plt.hist(above_average_reading["reading score"], bins = 10, color = "teal", edgecolor = "black")
plt.xlabel("Reading Score")
plt.ylabel("Frequency")
plt.title("Distribution of Students Scoring Above Average in Reading")
plt.show()
```

```python
print("Graph of students who scored above average in writing")
plt.hist(above_average_writing["writing score"], bins = 10, color = "peachpuff", edgecolor = "black")
plt.xlabel("Writing Score")
plt.ylabel("Frequency")
plt.title("Distribution of Students Scoring Above Average in Writing")
plt.show()
```

```python
print("What is the distribution of students who scored above average in all 3 subjects?")
colors = ["lightblue", "teal", "peachpuff"]

plt.hist([above_average_all["math score"], above_average_all["reading score"], above_average_all["writing score"]], bins = 10, color = colors, edgecolor = "black")
plt.xlabel("Score")
plt.ylabel("Frequency")
plt.title("Distribution of Students Scoring Above Average in all 3 subjects")
plt.show()


```

```python
print("How many students of each gender completed or did not complete the test preparation course?")

gender_prep = pd.crosstab(
    df["gender"],
    df["test preparation course"]
)

gender_prep.plot(kind="bar", figsize=(9, 5), edgecolor = "black")

plt.title("Gender Distribution by Test Preparation Course Status")
plt.xlabel("Gender")
plt.ylabel("Number of Students")
plt.xticks(rotation=0)
plt.legend(title="Test Preparation Course")

plt.tight_layout()
plt.show()

```

```python
print("What percentage of female students completed the test preparation course?")

female_students = df[df["gender"] == "female"]

female_completed = female_students[
    female_students["test preparation course"] == "completed"
]

female_percentage = (len(female_completed) / len(female_students)) * 100

print(f"Total female students: {len(female_students)}")
print(f"Female students who completed: {len(female_completed)}")
print(f"Completion percentage: {female_percentage:.2f}%")
```

```python
print("What percentage of male students completed the test preparation course?")

male_students = df[df["gender"] == "male"]

male_completed = male_students[
    male_students["test preparation course"] == "completed"
]

male_percentage = (len(male_completed) / len(male_students)) * 100

print(f"Total male students: {len(male_students)}")
print(f"Male students who completed: {len(male_completed)}")
print(f"Completion percentage: {male_percentage:.2f}%")
```

```python
print("What percentage of students belong to each race/ethnicity group?")

race_counts = df["race/ethnicity"].value_counts()

race_percentages = (race_counts / len(df)) * 100

print(race_percentages)
colors = ["lightblue", "lightgreen", "lightcoral", "lightsalmon", "lightpink"]
plt.bar(race_percentages.index, race_percentages.values, color=colors, edgecolor = "black")

plt.title("Percentage of Students by Race/Ethnicity")
plt.xlabel("Race/Ethnicity Group")
plt.ylabel("Percentage of Students (%)")

plt.show()
```

```python
print("How many and what percentages of female students belong to each race/ethnicity group?")

female_students = df[df["gender"] == "female"]

female_race_counts = female_students["race/ethnicity"].value_counts().sort_index()

print(female_race_counts)

female_race_percentages = (
    female_race_counts / len(female_students)
) * 100

print(female_race_percentages)
```

```python
print("How many and what percentages of male students belong to each race/ethnicity group?")

male_students = df[df["gender"] == "male"]

male_race_counts = male_students["race/ethnicity"].value_counts().sort_index()

print(male_race_counts)

male_race_percentages = (
    male_race_counts / len(male_students)
) * 100

print(male_race_percentages)
```

```python
print("How many students have parents with each education level?")

parent_education_counts = df["parental level of education"].value_counts()

plt.figure(figsize=(12, 6))
colors = ["blue", "green", "coral", "lightcoral", "pink", "yellow"]
plt.bar(parent_education_counts.index, parent_education_counts.values, color=colors, edgecolor = "black")

plt.title("Number of Students by Parental Education Level")
plt.xlabel("Parental Education Level")
plt.ylabel("Number of Students")

plt.xticks(rotation=45, ha="right")

plt.tight_layout()
plt.show()
```

```python
print("How do average student scores differ by parental education level?")

parent_education_averages = df.groupby(
    "parental level of education"
)[score_columns].mean()

print(parent_education_averages)
```

```python
print("How many students in each parental education group completed or did not complete test preparation?")

parent_prep_counts = pd.crosstab(
    df["parental level of education"],
    df["test preparation course"]
)

print(parent_prep_counts)
```

```python
print("How does test preparation completion vary across parental education groups?")

colors = [ "peachpuff", "salmon"]
parent_prep_counts.plot(kind="bar", figsize=(12, 6), color=colors, edgecolor = "black")
plt.title("Test Preparation Course by Parental Education Level")
plt.xlabel("Parental Education Level")
plt.ylabel("Number of Students")
plt.xticks(rotation=45, ha="right") 
plt.legend(title="Test Preparation")

plt.tight_layout()
plt.show()
```

```python
print("How does lunch type vary across parental education groups?")

lunch_parent_counts = pd.crosstab(
    df["parental level of education"],
    df["lunch"]
)

print(lunch_parent_counts)
```

```python
lunch_parent_counts.plot(kind="bar", figsize=(12, 6))

plt.title("Lunch Type by Parental Education Level")
plt.xlabel("Parental Education Level")
plt.ylabel("Number of Students")
plt.xticks(rotation=45, ha="right")
plt.legend(title="Lunch Type")
plt.tight_layout()
plt.show()
```

```python
print("What percentage of students in each parental education group receive each lunch type?")

lunch_parent_percentages = pd.crosstab(
    df["parental level of education"],
    df["lunch"],
    normalize="index"
) * 100

print(lunch_parent_percentages.round(2))
```

```python
lunch_parent_percentages.plot(
    kind="bar",
    figsize=(12, 6)
)

plt.title("Percentage of Lunch Types by Parental Education Level")
plt.xlabel("Parental Education Level")
plt.ylabel("Percentage of Students")
plt.xticks(rotation=45, ha="right")
plt.legend(title="Lunch Type")
plt.tight_layout()
plt.show()
```
